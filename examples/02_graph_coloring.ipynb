Three-coloring of a random 3-regular graph.

```python
import matplotlib.pyplot as plt
import networkx as nx

import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

```python
g = nx.random_regular_graph(d=3, n=40, seed=0)
problem = qqa.Coloring(g, num_category=3)
```

```python
result = qqa.anneal(problem, sol_size=100, num_epochs=2000, verbose=False)
print(f'conflict count: {int(result.best_obj)}')
```

```python
viz.plot_best_trajectory(result, show=False);
```
