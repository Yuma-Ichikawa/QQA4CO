Max-Cut on an Erdős–Rényi graph.

```python
# Install QQA on Google Colab (no-op if already installed locally).
# We prefer the released wheel on PyPI; users who want bleeding-edge
# ``main`` can set QQA_INSTALL_FROM_GIT=1 before running this cell.
import importlib.util
import os
import subprocess
import sys

if importlib.util.find_spec("qqa") is None:
    if os.environ.get("QQA_INSTALL_FROM_GIT") == "1":
        spec = "qqa @ git+https://github.com/Yuma-Ichikawa/QQA4CO.git"
    else:
        spec = "qqa"
    subprocess.check_call([sys.executable, "-m", "pip", "install", "--quiet", spec])
```

```python
import matplotlib.pyplot as plt  # noqa: F401
import networkx as nx

import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print("QQA version:", qqa.__version__)
```

```python
g = nx.erdos_renyi_graph(n=60, p=0.15, seed=0)
for u, v in g.edges:
    g[u][v]["weight"] = 1.0
problem = qqa.MaxCut(g)
```

```python
result = qqa.anneal(problem, sol_size=100, num_epochs=1500, verbose=False)
cut = -float(result.best_obj) / 2
print(f"approx cut size: {cut:.2f}")
```

```python
viz.plot_history(result, show=False);
```
