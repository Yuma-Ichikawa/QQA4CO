Max-Cut on an Erdős–Rényi graph.

```python
import matplotlib.pyplot as plt
import networkx as nx

import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

```python
g = nx.erdos_renyi_graph(n=60, p=0.15, seed=0)
for u, v in g.edges:
    g[u][v]['weight'] = 1.0
problem = qqa.MaxCut(g)
```

```python
result = qqa.anneal(problem, sol_size=100, num_epochs=1500, verbose=False)
cut = -float(result.best_obj) / 2
print(f'approx cut size: {cut:.2f}')
```

```python
viz.plot_history(result, show=False);
```
