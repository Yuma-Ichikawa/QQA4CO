# QQA 06 – Binary perceptron

Teacher/student binary perceptron learning as an optimization problem.

```python
import networkx as nx
import matplotlib.pyplot as plt
import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

## Discrete (binary) perceptron

Teacher–student setup: the teacher generates labels, the student must find binary weights that reproduce them.

```python
problem = qqa.BinaryPerceptron(N=50, alpha=0.5, seed=0, sharpness=10.0)
print(f'patterns: {problem.num_patterns}')
```

```python
result = qqa.anneal(problem, sol_size=200, num_epochs=2000, verbose=False)
s_best = result.best_sol[0]
print(f'surrogate loss: {result.best_obj:.3f}')
print(f'exact errors : {int(problem.error_count(s_best.unsqueeze(0)))}')
```

```python
viz.plot_best_trajectory(result, show=False);
```
