# QQA 05 – Sherrington–Kirkpatrick

Mean-field spin-glass ground-state energy.

```python
import networkx as nx
import matplotlib.pyplot as plt
import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

## Sherrington–Kirkpatrick model

All-to-all couplings $J_{ij}\sim\mathcal{N}(0, 1/N)$. The Parisi ground-state energy density is $e_0 \approx -0.7632$.

```python
problem = qqa.SherringtonKirkpatrick(N=100, seed=0)
result = qqa.anneal(problem, sol_size=200, num_epochs=2000, verbose=False)
print(f'E_0 / N = {result.best_obj / 100:.4f}  (target ≈ -0.7632)')
```

```python
viz.plot_history(result, show=False);
```
