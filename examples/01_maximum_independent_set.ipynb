## Setup

```python
import matplotlib.pyplot as plt
import networkx as nx

import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

## Problem: Maximum Independent Set on a random regular graph

```python
g = nx.random_regular_graph(d=3, n=50, seed=0)
problem = qqa.MaximumIndependentSet(g, penalty=2)
print(f'N = {problem.num_nodes}, |E| = {g.number_of_edges()}')
```

## Run QQA

```python
result = qqa.anneal(problem, sol_size=100, num_epochs=1500, verbose=False)
print(f'MIS size (lower bound): {-int(result.best_obj)}')
print(f'runtime: {result.runtime:.2f}s')
```

## Visualise the dynamics

```python
viz.plot_history(result, show=False);
```
