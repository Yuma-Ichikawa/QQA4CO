## Hopfield associative memory

Store $P$ random $\pm 1$ patterns via Hebbian couplings and check that QQA recovers a stored pattern.

```python
problem = qqa.HopfieldMemory(N=80, patterns=3, seed=0)
result = qqa.anneal(problem, sol_size=128, num_epochs=1500, verbose=False)
print(f"energy: {result.best_obj:.3f}  (target ≈ {-80 / 2:.3f} per pattern)")
```

```python
s_best = result.best_sol.unsqueeze(0)
m = problem.overlap(s_best)
print(f"overlap with stored patterns: {m[0].tolist()}")
```
