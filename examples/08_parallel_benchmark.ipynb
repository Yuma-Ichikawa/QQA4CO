# QQA 08 – Parallel benchmark

Sweep QQA hyper-parameters and visualise the outcome.

```python
import networkx as nx
import matplotlib.pyplot as plt
import qqa
from qqa import visualization as viz

qqa.fix_seed(0)
print('QQA version:', qqa.__version__)
```

## Parallel benchmark with hyper-parameter sweep

Run multiple QQA configurations on the same problem and inspect the parallel-coordinates plot of the results.

```python
import itertools, pandas as pd
problem = qqa.SherringtonKirkpatrick(N=80, seed=0)
rows = []
for mb, Mb, dp in itertools.product([-3, -2, -1], [0.0, 0.1], [0.0, 0.1]):
    r = qqa.anneal(problem, sol_size=64, num_epochs=600,
                   min_bg=mb, max_bg=Mb, div_param=dp, verbose=False)
    rows.append({'min_bg': mb, 'max_bg': Mb, 'div_param': dp,
                 'best_obj_per_N': float(r.best_obj) / 80})
df = pd.DataFrame(rows)
df
```

```python
try:
    fig = viz.plot_parallel_coordinates(df, objective='best_obj_per_N', show=False)
    fig
except Exception as e:
    print('Plotly not installed; install with pip install qqa[plotly]')
    print(e)
```
