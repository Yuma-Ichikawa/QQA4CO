## 3D Edwards–Anderson spin glass

A $L^3$ cubic lattice with independent $\mathcal{N}(0, 1)$ couplings on nearest-neighbour bonds.

```python
problem = qqa.EdwardsAnderson(L=4, dim=3, seed=0)
print(f'N = {problem.num_spins}, couplings shape = {tuple(problem.J.shape)}')
```

```python
result = qqa.anneal(problem, sol_size=200, num_epochs=2000, verbose=False)
print(f'E_0 / N ≈ {result.best_obj / problem.num_spins:.4f}')
```

```python
viz.plot_best_trajectory(result, show=False);
```
